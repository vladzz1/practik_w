```python
from IPython.display import display, Markdown

votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
apple_count = votes.count("apple")
banana_count = votes.count("banana")
orange_count = votes.count("orange")
data = [
    ("Яблуко", apple_count),
    ("Банан", banana_count),
    ("Апельсин", orange_count)
]
sorted_data = sorted(data, key=lambda x: x[1], reverse=True)

print("фрукт лідер:", sorted_data[0][0])

display(Markdown(f"""
| фрукт | кількість |
|---|---|
| яблука | {apple_count} |
| банани | {banana_count} |
| апельсини | {orange_count} |
"""))

display(Markdown(f"""
| місце | фрукт |
|---|---|
| 1 | {sorted_data[0][0]} |
| 2 | {sorted_data[1][0]} |
| 3 | {sorted_data[2][0]} |
"""))
```
