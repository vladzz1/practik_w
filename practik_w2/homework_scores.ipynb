```python
from IPython.display import display, Markdown

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

math_average = sum(math) / len(math)
math_maximum = max(math)
math_minimum = min(math)

physics_average = sum(physics) / len(physics)
physics_maximum = max(physics)
physics_minimum = min(physics)

english_average = sum(english) / len(english)
english_maximum = max(english)
english_minimum = min(english)

data = [
    ("math", math_average),
    ("physics", physics_average),
    ("english", english_average)
]
sorted_data = sorted(data, key=lambda x: x[1], reverse=True)

display(Markdown(f"""
| предмет | середній бал |
|---|---|
| math | {math_average} |
| physics | {physics_average} |
| english | {english_average} |
"""))

display(Markdown(f"""
| предмет | мінімальний бал |
|---|---|
| math | {math_minimum} |
| physics | {physics_minimum} |
| english | {english_minimum} |
"""))

display(Markdown(f"""
| предмет | максимальний бал |
|---|---|
| math | {math_maximum} |
| physics | {physics_maximum} |
| english | {english_maximum} |
"""))

print("предмет з найбільшим середнім балом:", sorted_data[0][0])
```
