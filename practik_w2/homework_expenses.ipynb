```python
from IPython.display import display, Markdown

costs = [1200, 1500, 800, 600, 1000, 400, 900]
total_amount = 0
number_days = 0
i = 0
while i < 7:
    total_amount += costs[i]
    if costs[i] > 1000:
        number_days += 1
    i += 1
average = sum(costs) / len(costs)
maximum = max(costs)
minimum = min(costs)

display(Markdown(f"""
| Показник | Значення |
|---|---|
| Сума витрат | {total_amount} |
| Середнє значення | {average:.2f} |
| Максимальне значення | {maximum} |
| Мінімальне значення | {minimum} |
| Кількість днів з витратами більшими за 1000 | {number_days} |
"""))
```
