# Звіт із продажів

# У цьому розділі ми рахуємо підсумки за простими даними

```python
sales = [120, 135, 100, 150, 90, 200, 170]
i = 0
generally = 0
maximum = sales[0]
minimum = sales[0]
while i < 7:
    generally += sales[i]
    if maximum < sales[i]:
        maximum = sales[i]
    if minimum > sales[i]:
        minimum = sales[i]
    i += 1
average = generally / 7
```
